## 用真实 API 验证数据可读

下面直接构造 `TianmoucDataReader`，确认测试样本可以被 TianmouCV 的数据读取入口正常读取。

```python
from tianmoucv.data import TianmoucDataReader

dataset = TianmoucDataReader(str(info.sample_dir), N=1, print_info=False)
print('样本数:', len(dataset))

sample = dataset[0]
print('sample keys:')
for key in sorted(sample.keys()):
    value = sample[key]
    if hasattr(value, 'shape'):
        print(f'  {key}: {tuple(value.shape)}')
    else:
        print(f'  {key}: {type(value).__name__}')
```

```python
plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
plt.imshow(sample['F0'].numpy())
plt.title('F0')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(sample['F0_HDR'].numpy())
plt.title('F0_HDR')
plt.axis('off')
plt.tight_layout()
plt.show()
```
